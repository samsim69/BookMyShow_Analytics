```python
import pandas as pd
import numpy as np
import random

random.seed(42)
np.random.seed(42)

customers = pd.read_csv("../Dataset/customers.csv")
shows = pd.read_csv("../Dataset/shows.csv")
```

```python
TOTAL_BOOKINGS = 500000
```

```python
payment_methods = ["UPI","Credit Card","Debit Card","Wallet","Net Banking"]
payment_weights = [50,20,15,10,5]
```

```python
seat_categories = ["Silver","Gold","Platinum","Recliner"]
seat_weights = [55,25,15,5]
```

```python
booking_status = ["Confirmed","Cancelled","Refunded"]
status_weights = [94,4,2]
```

```python
coupon_codes = [
    "NONE",
    "WELCOME100",
    "MOVIE50",
    "FESTIVE200",
    "FIRSTBOOK"
]

coupon_weights = [70,10,8,7,5]
```

```python
bookings = []

customer_ids = customers["customer_id"].values

show_ids = shows["show_id"].values
ticket_prices = shows["ticket_price"].values

for booking_id in range(1, TOTAL_BOOKINGS + 1):

    customer_id = np.random.choice(customer_ids)

    idx = np.random.randint(0, len(show_ids))

    show_id = show_ids[idx]
    ticket_price = ticket_prices[idx]

    seats = np.random.randint(1,5)

    base_amount = seats * ticket_price

    coupon = random.choices(
        coupon_codes,
        weights=coupon_weights,
        k=1
    )[0]

    discount_map = {
        "NONE":0,
        "WELCOME100":100,
        "MOVIE50":50,
        "FESTIVE200":200,
        "FIRSTBOOK":150
    }

    discount = discount_map[coupon]

    convenience_fee = round(base_amount * 0.03)

    gst = round((base_amount - discount) * 0.18)

    total = max(0, base_amount - discount + convenience_fee + gst)

    bookings.append({

        "booking_id": booking_id,

        "customer_id": customer_id,

        "show_id": show_id,

        "seats_booked": seats,

        "seat_category": random.choices(
            seat_categories,
            weights=seat_weights,
            k=1
        )[0],

        "payment_method": random.choices(
            payment_methods,
            weights=payment_weights,
            k=1
        )[0],

        "booking_status": random.choices(
            booking_status,
            weights=status_weights,
            k=1
        )[0],

        "coupon_code": coupon,

        "discount": discount,

        "convenience_fee": convenience_fee,

        "gst": gst,

        "total_amount": round(total,2)

    })

    if booking_id % 50000 == 0:
        print(f"{booking_id:,} bookings generated...")
```

```python
bookings_df = pd.DataFrame(bookings)

bookings_df.to_csv("../Dataset/bookings.csv", index=False)

print(bookings_df.shape)

bookings_df.head()
```
