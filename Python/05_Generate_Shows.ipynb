```python
import pandas as pd
import random
from datetime import datetime, timedelta

random.seed(42)

movies = pd.read_csv("../Dataset/movies.csv")
screens = pd.read_csv("../Dataset/screens.csv")
theatres = pd.read_csv("../Dataset/theatres.csv")
```

```python
show_timings = [

"09:00",

"12:00",

"15:30",

"18:30",

"21:45"

]
```

```python
base_prices = {

"2D":180,

"3D":280,

"IMAX":550,

"4DX":450

}
```

```python
city_multiplier = {

"Mumbai":1.35,

"Delhi":1.30,

"Bengaluru":1.28,

"Hyderabad":1.20,

"Chennai":1.18,

"Kolkata":1.10,

"Pune":1.15,

"Ahmedabad":1.08,

"Jaipur":1.00,

"Lucknow":0.95,

"Surat":1.00,

"Indore":0.95,

"Nagpur":0.90,

"Bhopal":0.90,

"Patna":0.85,

"Kanpur":0.85,

"Kochi":1.05,

"Coimbatore":1.00,

"Visakhapatnam":1.00,

"Chandigarh":1.10

}
```

```python
shows = []

show_id = 1

start_date = datetime(2025, 1, 1)

end_date = datetime(2025, 12, 31)

date_pool = pd.date_range(start_date, end_date)

# fast lookup

theatre_lookup = theatres.set_index("theatre_id")["city"].to_dict()

# convert dataframe → lists (VERY IMPORTANT FOR SPEED)

screens_list = screens.to_dict("records")

movies_list = movies.to_dict("records")

# keep dataset size similar to original logic

target_size = len(screens_list) * 30 * 5

for _ in range(target_size):

    current_date = random.choice(date_pool)

    screen = random.choice(screens_list)

    movie = random.choice(movies_list)

    city = theatre_lookup[screen["theatre_id"]]

    price = base_prices[screen["screen_type"]]

    price *= city_multiplier[city]

    if current_date.weekday() >= 5:

        price *= 1.20

    shows.append({

        "show_id": show_id,

        "movie_id": movie["movie_id"],

        "screen_id": screen["screen_id"],

        "show_date": current_date.date(),

        "show_time": random.choice(show_timings),

        "ticket_price": round(price)

    })

    show_id += 1
```

```python
shows_df = pd.DataFrame(shows)

shows_df.head()
```

```python
shows_df.to_csv(

"../Dataset/shows.csv",

index=False

)

print("Shows Dataset Created Successfully!")
```
