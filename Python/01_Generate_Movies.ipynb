```python
import os
import sys

# Go one level up (Project Folder)
project_root = os.path.abspath("..")

# Add project folder to Python path
if project_root not in sys.path:
    sys.path.append(project_root)

import pandas as pd
import random
from datetime import datetime, timedelta

from Utilities.config import TOTAL_MOVIES

print("Project Root:", project_root)
print("Import Successful!")
```

```python
genres = [
    "Action",
    "Comedy",
    "Drama",
    "Crime",
    "Thriller",
    "Romance",
    "Sci-Fi",
    "Animation",
    "Family",
    "Horror"
]

languages = [
    "Hindi",
    "English",
    "Tamil",
    "Telugu",
    "Kannada",
    "Malayalam"
]

certificates = ["U", "UA", "A"]

statuses = [
    "Running",
    "Upcoming",
    "Completed"
]
```

```python
production_houses = [

"Dharma Productions",

"Yash Raj Films",

"T-Series",

"Red Chillies Entertainment",

"Hombale Films",

"Marvel Studios",

"Warner Bros",

"Universal Pictures",

"Disney",

"Mythri Movie Makers"

]
```

```python
directors = [

"Rajkumar Hirani",

"Sandeep Reddy Vanga",

"Christopher Nolan",

"James Cameron",

"Atlee",

"Sukumar",

"Lokesh Kanagaraj",

"Prashanth Neel",

"Greta Gerwig",

"Shawn Levy",

"Nag Ashwin",

"S. S. Rajamouli"

]
```

```python
actors = [

"Shah Rukh Khan",

"Ranbir Kapoor",

"Prabhas",

"Allu Arjun",

"Vijay",

"Yash",

"Ram Charan",

"Tom Cruise",

"Ryan Reynolds",

"Cillian Murphy",

"Margot Robbie",

"Keanu Reeves"

]
```

```python
adjectives = [

"Shadow",

"Final",

"Dark",

"Lost",

"Hidden",

"Last",

"Golden",

"Silent",

"Broken",

"Eternal",

"Royal",

"Infinite"

]

nouns = [

"Empire",

"Mission",

"War",

"Legacy",

"Kingdom",

"Journey",

"Dream",

"Code",

"Revenge",

"Storm",

"Chronicles",

"Destiny"

]
```

```python
movies = []

for movie_id in range(1, TOTAL_MOVIES + 1):

    title = f"{random.choice(adjectives)} {random.choice(nouns)}"

    genre = random.choice(genres)

    language = random.choice(languages)

    release_date = datetime(
        random.randint(2023, 2025),
        random.randint(1, 12),
        random.randint(1, 28)
    )

    ott_release = release_date + timedelta(days=60)

    budget = random.randint(30, 400)

    box_office = round(budget * random.uniform(0.6, 6), 2)

    rating = round(random.uniform(5.5, 9.5), 1)

    runtime = random.randint(95, 190)

    movies.append({

        "movie_id": movie_id,

        "movie_name": title,

        "genre": genre,

        "language": language,

        "director": random.choice(directors),

        "lead_actor": random.choice(actors),

        "runtime": runtime,

        "release_date": release_date.date(),

        "ott_release": ott_release.date(),

        "budget_cr": budget,

        "box_office_cr": box_office,

        "imdb_rating": rating,

        "certificate": random.choice(certificates),

        "production_house": random.choice(production_houses),

        "status": random.choice(statuses)

    })
```

```python
movies_df = pd.DataFrame(movies)
movies_df.head()
```

```python
movies_df.to_csv("../Dataset/movies.csv", index=False)

print("Movies dataset generated successfully!")
```
