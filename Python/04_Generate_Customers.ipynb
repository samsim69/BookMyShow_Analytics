```python
import pandas as pd
import random
from faker import Faker

fake = Faker("en_IN")

random.seed(42)
Faker.seed(42)

TOTAL_CUSTOMERS = 100000
```

```python
cities = [
    "Mumbai",
    "Delhi",
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Kolkata",
    "Pune",
    "Ahmedabad",
    "Jaipur",
    "Lucknow",
    "Surat",
    "Indore",
    "Nagpur",
    "Patna",
    "Bhopal",
    "Kochi",
    "Coimbatore",
    "Visakhapatnam",
    "Chandigarh",
    "Kanpur"
]
```

```python
occupations = {

    "Student":22,

    "Software Engineer":12,

    "Business":10,

    "Teacher":6,

    "Doctor":5,

    "Government Employee":6,

    "Private Employee":18,

    "Freelancer":5,

    "Retired":4,

    "Self Employed":7,

    "Other":5

}
```

```python
membership = {

    "Regular":72,

    "Gold":22,

    "Platinum":6

}
```

```python
genres = [

"Action",

"Drama",

"Comedy",

"Crime",

"Thriller",

"Romance",

"Sci-Fi",

"Horror",

"Animation"

]
```

```python
languages = [

"Hindi",

"English",

"Tamil",

"Telugu",

"Kannada",

"Malayalam"

]
```

```python
customers = []

for customer_id in range(1, TOTAL_CUSTOMERS + 1):

    age = random.randint(18, 65)

    customers.append({

        "customer_id": customer_id,

        "customer_name": fake.name(),

        "gender": random.choice(["Male", "Female"]),

        "age": age,

        "city": random.choice(cities),

        "occupation": random.choices(
            population=list(occupations.keys()),
            weights=list(occupations.values()),
            k=1
        )[0],

        "membership": random.choices(
            population=list(membership.keys()),
            weights=list(membership.values()),
            k=1
        )[0],

        "preferred_genre": random.choice(genres),

        "preferred_language": random.choice(languages),

        "monthly_movie_budget": random.randint(300,3000),

        "food_purchase_probability": round(random.uniform(0.20,0.90),2),

        "app_rating": round(random.uniform(3.0,5.0),1)

    })
```

```python
customers_df = pd.DataFrame(customers)

customers_df.head()
```

```python
customers_df.to_csv(

"../Dataset/customers.csv",

index=False

)

print("Customers Dataset Created Successfully!")
```
