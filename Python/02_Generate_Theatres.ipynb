```python
import pandas as pd
import random

random.seed(42)
```

```python
cities = {
    "Mumbai": "Maharashtra",
    "Delhi": "Delhi",
    "Bengaluru": "Karnataka",
    "Hyderabad": "Telangana",
    "Chennai": "Tamil Nadu",
    "Kolkata": "West Bengal",
    "Pune": "Maharashtra",
    "Ahmedabad": "Gujarat",
    "Jaipur": "Rajasthan",
    "Lucknow": "Uttar Pradesh",
    "Surat": "Gujarat",
    "Indore": "Madhya Pradesh",
    "Nagpur": "Maharashtra",
    "Bhopal": "Madhya Pradesh",
    "Patna": "Bihar",
    "Kanpur": "Uttar Pradesh",
    "Kochi": "Kerala",
    "Coimbatore": "Tamil Nadu",
    "Chandigarh": "Chandigarh",
    "Visakhapatnam": "Andhra Pradesh"
}
```

```python
brands = [
    "PVR",
    "INOX",
    "Cinepolis",
    "Miraj",
    "Carnival",
    "Asian",
    "Mukta A2",
    "Wave",
    "MovieMax",
    "Movietime"
]
```

```python
screen_types = [
    "2D",
    "3D",
    "IMAX",
    "4DX"
]
```

```python
theatres = []

for theatre_id in range(1,101):

    city = random.choice(list(cities.keys()))

    state = cities[city]

    brand = random.choice(brands)

    theatre_name = f"{brand} {city}"

    screen_count = random.randint(3,10)

    capacity = screen_count * random.randint(120,250)

    premium = random.choice(["Yes","No"])

    opening_year = random.randint(2005,2024)

    theatres.append({

        "theatre_id":theatre_id,

        "theatre_name":theatre_name,

        "city":city,

        "state":state,

        "screen_count":screen_count,

        "capacity":capacity,

        "premium":premium,

        "opening_year":opening_year

    })
```

```python
theatres_df = pd.DataFrame(theatres)

theatres_df.head()
```

```python
theatres_df.to_csv("../Dataset/theatres.csv",index=False)

print("Theatres Dataset Created Successfully!")
```
