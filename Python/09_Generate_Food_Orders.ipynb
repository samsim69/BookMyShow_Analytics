```python
import pandas as pd
import numpy as np
import random

random.seed(42)
np.random.seed(42)

bookings = pd.read_csv("../Dataset/bookings.csv")
customers = pd.read_csv("../Dataset/customers.csv")
```

```python
food = bookings.merge(
    customers[
        ["customer_id","food_purchase_probability"]
    ],
    on="customer_id",
    how="left"
)
```

```python
menu = {

"Popcorn":250,

"Nachos":220,

"Coke":150,

"Pepsi":150,

"Burger":320,

"Pizza":450,

"French Fries":180,

"Combo Meal":650

}
```

```python
orders = []

food_order_id = 1

for _, row in food.iterrows():

    probability = row["food_purchase_probability"]

    if random.random() <= probability:

        quantity = np.random.randint(1,4)

        item = random.choice(list(menu.keys()))

        price = menu[item]

        orders.append({

            "food_order_id": food_order_id,

            "booking_id": row["booking_id"],

            "food_item": item,

            "quantity": quantity,

            "unit_price": price,

            "total_price": quantity * price

        })

        food_order_id += 1
```

```python
food_orders_df = pd.DataFrame(orders)

food_orders_df.head()
```

```python
food_orders_df.to_csv(
    "../Dataset/food_orders.csv",
    index=False
)

print(food_orders_df.shape)
```
