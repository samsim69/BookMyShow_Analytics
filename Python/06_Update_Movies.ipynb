```python
import pandas as pd
import random

random.seed(42)

movies = pd.read_csv("../Dataset/movies.csv")
```

```python
popularity = []

for _, movie in movies.iterrows():

    rating = movie["imdb_rating"]

    if rating >= 9:
        score = random.randint(90,100)

    elif rating >= 8:
        score = random.randint(75,89)

    elif rating >= 7:
        score = random.randint(55,74)

    else:
        score = random.randint(30,54)

    popularity.append(score)

movies["popularity_score"] = popularity
```

```python
movies.to_csv("../Dataset/movies.csv", index=False)

movies.head()
```
