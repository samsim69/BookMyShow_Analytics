```python
from sqlalchemy import create_engine

engine = create_engine(
    "mysql+pymysql://root:s%40msim69@127.0.0.1:3306/bookmyshow_analytics"
)
```

```python
import pandas as pd

movies = pd.read_csv("../Dataset/movies.csv")
theatres = pd.read_csv("../Dataset/theatres.csv")
customers = pd.read_csv("../Dataset/customers.csv")
screens = pd.read_csv("../Dataset/screens.csv")
shows = pd.read_csv("../Dataset/shows.csv")
bookings = pd.read_csv("../Dataset/bookings.csv")
food_orders = pd.read_csv("../Dataset/food_orders.csv")
reviews = pd.read_csv("../Dataset/reviews.csv")

print("All CSVs loaded successfully!")
```

```python
movies["release_date"] = pd.to_datetime(movies["release_date"])
movies["ott_release"] = pd.to_datetime(movies["ott_release"])

shows["show_date"] = pd.to_datetime(shows["show_date"])
```

```python
movies.to_sql(
    "movies",
    engine,
    if_exists="append",
    index=False
)

print("Movies Imported")
```

```python
theatres.to_sql(
    "theatres",
    engine,
    if_exists="append",
    index=False
)

print("Theatres Imported")
```

```python
customers.to_sql(
    "customers",
    engine,
    if_exists="append",
    index=False
)

print("Customers Imported")
```

```python
screens.to_sql(
    "screens",
    engine,
    if_exists="append",
    index=False
)

print("Screens Imported")
```

```python
shows.to_sql(
    "shows",
    engine,
    if_exists="append",
    index=False
)

print("Shows Imported")
```

```python
bookings.to_sql(
    "bookings",
    engine,
    if_exists="append",
    index=False,
    chunksize=10000,
    method="multi"
)

print("Bookings Imported")
```

```python
food_orders.to_sql(
    "food_orders",
    engine,
    if_exists="append",
    index=False,
    chunksize=10000,
    method="multi"
)

print("Food Orders Imported")
```

```python
reviews.to_sql(
    "reviews",
    engine,
    if_exists="append",
    index=False,
    chunksize=5000,
    method="multi"
)

print("Reviews Imported")
```

```python
from sqlalchemy import text

tables = [
    "movies",
    "theatres",
    "customers",
    "screens",
    "shows",
    "bookings",
    "food_orders",
    "reviews"
]

with engine.connect() as conn:
    for table in tables:
        result = conn.execute(text(f"SELECT COUNT(*) FROM {table}"))
        print(f"{table}: {result.scalar()} rows")
```
