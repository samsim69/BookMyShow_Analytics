```python
import pandas as pd

shows = pd.read_csv("../Dataset/shows.csv")
screens = pd.read_csv("../Dataset/screens.csv")
movies = pd.read_csv("../Dataset/movies.csv")
```

```python
shows = shows.merge(
    screens[["screen_id", "seat_capacity"]],
    on="screen_id",
    how="left"
)
```

```python
shows = shows.merge(
    movies[["movie_id", "popularity_score"]],
    on="movie_id",
    how="left"
)
```

```python
import numpy as np

np.random.seed(42)

occupancy = []

for _, row in shows.iterrows():

    base = row["popularity_score"] * 0.7

    weekend_bonus = 15 if pd.to_datetime(row["show_date"]).weekday() >= 5 else 0

    evening_bonus = 10 if row["show_time"] in ["18:30", "21:45"] else 0

    value = base + weekend_bonus + evening_bonus

    value = max(20, min(98, value))

    occupancy.append(value)

shows["occupancy_percent"] = occupancy
```

```python
shows["tickets_sold"] = (
    shows["seat_capacity"] *
    shows["occupancy_percent"] / 100
).astype(int)
```

```python
shows.to_csv("../Dataset/shows.csv", index=False)

shows.head()
```
