```python
import pandas as pd
import random

random.seed(42)

theatres = pd.read_csv("../Dataset/theatres.csv")
```

```python
screen_types = {

    "2D":0.60,

    "3D":0.22,

    "IMAX":0.10,

    "4DX":0.08

}
```

```python
screens = []

screen_id = 1

for _, theatre in theatres.iterrows():

    for number in range(1, theatre["screen_count"] + 1):

        screen_type = random.choices(

            population=list(screen_types.keys()),

            weights=list(screen_types.values()),

            k=1

        )[0]

        if screen_type == "2D":
            seats = random.randint(180, 260)

        elif screen_type == "3D":
            seats = random.randint(170, 240)

        elif screen_type == "IMAX":
            seats = random.randint(280, 420)

        else:
            seats = random.randint(150, 220)

        screens.append({

            "screen_id": screen_id,

            "theatre_id": theatre["theatre_id"],

            "screen_number": number,

            "screen_type": screen_type,

            "seat_capacity": seats

        })

        screen_id += 1
```

```python
screens_df = pd.DataFrame(screens)

screens_df.head()
```

```python
screens_df.to_csv("../Dataset/screens.csv", index=False)

print("Screens dataset created successfully!")
```
