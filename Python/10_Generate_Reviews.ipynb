```python
import pandas as pd
import numpy as np
import random

random.seed(42)
np.random.seed(42)

movies = pd.read_csv("../Dataset/movies.csv")
shows = pd.read_csv("../Dataset/shows.csv")
bookings = pd.read_csv("../Dataset/bookings.csv")
```

```python
reviews_base = bookings.merge(
    shows[["show_id","movie_id"]],
    on="show_id",
    how="left"
)

reviews_base = reviews_base.merge(
    movies[["movie_id","imdb_rating"]],
    on="movie_id",
    how="left"
)
```

```python
reviews_base = reviews_base[
    reviews_base["booking_status"] == "Confirmed"
]
```

```python
positive_comments = [
    "Amazing movie",
    "Loved every minute",
    "Must watch",
    "Excellent performances",
    "Highly recommended"
]

neutral_comments = [
    "Average movie",
    "Good but not great",
    "Decent watch",
    "One time watch",
    "Okay experience"
]

negative_comments = [
    "Not worth it",
    "Poor screenplay",
    "Disappointing",
    "Could have been better",
    "Waste of time"
]
```

```python
reviews = []

review_id = 1

for _, row in reviews_base.iterrows():

    imdb = row["imdb_rating"]

    probability = 0.20

    if random.random() > probability:
        continue

    if imdb >= 8.5:
        rating = random.choice([4,5,5,5])

    elif imdb >= 7.5:
        rating = random.choice([3,4,4,5])

    elif imdb >= 6.5:
        rating = random.choice([3,3,4])

    else:
        rating = random.choice([1,2,2,3])

    if rating >= 4:
        comment = random.choice(positive_comments)

    elif rating == 3:
        comment = random.choice(neutral_comments)

    else:
        comment = random.choice(negative_comments)

    reviews.append({

        "review_id": review_id,

        "booking_id": row["booking_id"],

        "movie_id": row["movie_id"],

        "user_rating": rating,

        "review_comment": comment

    })

    review_id += 1
```

```python
reviews_df = pd.DataFrame(reviews)

reviews_df.head()
```

```python
reviews_df.to_csv(
    "../Dataset/reviews.csv",
    index=False
)

print(reviews_df.shape)
```
